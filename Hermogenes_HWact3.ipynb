```python
import numpy as np
import pandas as pd

customers = ['A', 'B', 'C', 'D', 'E']

x1 = np.array([1, 2, 3, 5, 6], dtype=float)
x2 = np.array([4, 3, 7, 2, 6], dtype=float)
y = np.array([0, 0, 1, 1, 1], dtype=float)

m1 = 0.8
m2 = 0.4
b = -4.0

eta = 0.1

print("Dataset and parameters loaded successfully!")
print("m1 =", m1)
print("m2 =", m2)
print("b  =", b)
print("Learning rate =", eta)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print("Sigmoid function ready!")
```

```python
z = m1*x1 + m2*x2 + b
y_pred = sigmoid(z)

task1 = pd.DataFrame({
    'Customer': customers,
    'x1 Time': x1,
    'x2 Pages': x2,
    'y Actual': y,
    'z Score': z,
    'Predicted Probability': y_pred
})

display(task1.round(6))
```

```python
loss = -(y * np.log(y_pred) + (1-y) * np.log(1-y_pred))

average_loss = np.mean(loss)

task2 = pd.DataFrame({
    'Customer': customers,
    'x1 Time': x1,
    'x2 Pages': x2,
    'y Actual': y,
    'Predicted Probability': y_pred,
    'Loss': loss
})

display(task2.round(6))

print("Average BCE Loss =", round(average_loss, 6))
```

```python
dm1 = np.mean((y_pred - y) * x1)
dm2 = np.mean((y_pred - y) * x2)
db = np.mean(y_pred - y)

print("dL/dm1 =", dm1)
print("dL/dm2 =", dm2)
print("dL/db =", db)

new_m1 = m1 - eta * dm1
new_m2 = m2 - eta * dm2
new_b = b - eta * db

print("New m1 =", new_m1)
print("New m2 =", new_m2)
print("New b =", new_b)
```

```python
new_z = new_m1*x1 + new_m2*x2 + new_b
new_y_pred = sigmoid(new_z)

task4 = pd.DataFrame({
    'Customer': customers,
    'x1 Time': x1,
    'x2 Pages': x2,
    'y Actual': y,
    'New Probability': new_y_pred
})

display(task4.round(6))
```

```python
new_loss = -(y * np.log(new_y_pred) +
             (1-y) * np.log(1-new_y_pred))

new_average_loss = np.mean(new_loss)

task5 = pd.DataFrame({
    'Customer': customers,
    'New Probability': new_y_pred,
    'New Loss': new_loss
})

display(task5.round(6))

print("New Average BCE Loss =", round(new_average_loss, 6))
```
